```python
plt.ion()
xs,ys = numpy.linspace(-10, 10, 200),numpy.linspace(-10, 10, 200)
X,Y = numpy.meshgrid(xs,ys)
plt.figure()
plt.contourf(X,Y, dummy_func(X, Y, 0.2,1))

X_amr,Y_amr = X[::10,::10] ,Y[::10,::10]
plt.figure()
plt.contourf(X_amr,Y_amr, dummy_func(X_amr, Y_amr, 0.2,1))
plt.scatter(X_amr,Y_amr,s = 1)
```

```python
from netgen.geom2d import unit_square, MakeCircle, SplineGeometry
from netgen.meshing import Element0D, Element1D, Element2D, MeshPoint, FaceDescriptor, Mesh
from netgen.csg import Pnt
import ngsolve
quads =False# True
N=5

mesh = Mesh()
mesh.SetGeometry(unit_square)
mesh.dim = 2


pnums = []
for i in range(N + 1):
    for j in range(N + 1):
        pnums.append(mesh.Add(MeshPoint(Pnt(i / N, j / N, 0))))
idx_dom = mesh.AddRegion("mat", dim=2)
# mesh.SetMaterial(1, "mat")
for j in range(N):
    for i in range(N):
        if quads:
            mesh.Add(Element2D(1, [
                pnums[i + j * (N + 1)],                                   
                pnums[i + (j + 1) * (N + 1)],
                pnums[i + 1 + (j + 1) * (N + 1)],
                pnums[i + 1 + j * (N + 1)]
                                  ])
                    )
        else:
            mesh.Add(Element2D(1, [pnums[i + j * (N + 1)], pnums[i + (j + 1) * (N + 1)], pnums[i + 1 + j * (N + 1)]]))
            mesh.Add(Element2D(1, [pnums[i + (j + 1) * (N + 1)], pnums[i + 1 + (j + 1) * (N + 1)], pnums[i + 1 + j * (N + 1)]]))
mesh.Add (FaceDescriptor(surfnr=1,domin=1,bc=1))
ngsolvemesh = ngsolve.Mesh(mesh)
```
